# 取得したニュースに対して前処理を行う

```python
import pandas as pd

df = pd.read_csv("news.csv")
```

```python
df['本文']
```

```python
# 表示幅を制限なしに変更
pd.set_option('display.max_colwidth', None)

# '本文'カラムを表示
print(df['本文'])
```

```python
df['本文'] = df['本文'].str.replace(r'\n|\r', '', regex=True)
```

```python
df['本文'] = df['本文'].str.replace(r'　', '', regex=True)
```

```python
df.loc[60:69, '本文']
```

```python
import MeCab

tagger = MeCab.Tagger("-Owakati")  # 分かち書き用

# '本文'カラムを形態素解析
df['形態素解析'] = df['本文'].apply(lambda text: tagger.parse(text).strip())

df.loc[60:69, '形態素解析']
```

```python
import MeCab

# MeCabのTaggerを初期化（デフォルト辞書を使用）
tagger = MeCab.Tagger()

def extract_nouns(text):
    """
    入力されたテキストから名詞を抽出する関数
    """
    nouns = []
    parsed = tagger.parse(text)
    for line in parsed.splitlines():
        if line == 'EOS' or line == '':
            continue
        parts = line.split('\t')
        if len(parts) > 1:
            # 品詞情報を取得（形態素解析結果の第2カラム）
            features = parts[1].split(',')
            if features[0] == '名詞':  # 名詞だけを抽出
                nouns.append(parts[0])
    return ' '.join(nouns)

# '本文'カラムの名詞を抽出して新しい列に追加
df['名詞抽出'] = df['本文'].apply(extract_nouns)

# 結果を表示
df.loc[60:69, '名詞抽出']
```
