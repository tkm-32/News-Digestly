### pkeをTF-IDFアルゴリズムで使用

```python
import pke
import fugashi
from nltk.corpus import stopwords

# サンプルニュースデータ
text = """
AI技術は近年注目を集めており、医療や自動運転、金融などさまざまな分野で活用されています。
特にディープラーニングの進化によって、AIの能力が飛躍的に向上しています。
IoTデバイスとAIの連携も、これからのテクノロジーの重要な要素とされています。
"""

# MeCabを使った分かち書き関数
def preprocess_text_with_mecab(text):
    tagger = fugashi.Tagger()
    tokens = [word.surface for word in tagger(text) if word.feature.lemma != '']
    return " ".join(tokens)

# テキストを分かち書き
preprocessed_text = preprocess_text_with_mecab(text)

# カスタムストップワードリスト（日本語用）
stoplist = [
    "の", "に", "は", "を", "が", "と", "で", "も", "た", "する", "いる", "こと", "これ",
    "それ", "ため", "よう", "さん", "ある", "ない", "なる", "おり", "より", "られる"
]
# 以下がpkeを使用する際の流れ

# PKEのTF-IDFアルゴリズムを使用
extractor = pke.unsupervised.TfIdf()

# 分かち書きされたテキストをPKEに読み込む
extractor.load_document(input=preprocessed_text, language='ja', normalization=None)

# キーフレーズ候補を選択
extractor.candidate_selection()

# ストップワードを手動で除外
candidates = {key: value for key, value in extractor.candidates.items() if key not in stoplist}
extractor.candidates = candidates

# TF-IDFスコアを計算してキーフレーズを抽出
extractor.candidate_weighting()

# 上位のキーフレーズを取得
keyphrases = extractor.get_n_best(n=10)

# キーフレーズを表示
print("抽出されたキーフレーズ:")
for phrase, score in keyphrases:
    print(f"{phrase}: {score}")
```

**memo**
- 先ほどのtf-idfに比べると名詞が繋がっており上手く抽出できている印象

### TextRankアルゴリズムでキーワード抽出

### アルゴリズム上不要なトークンを除去する必要があり、先ほどの関数を改良した関数

```python
def tokenize_with_mecab_ver2(text):
    tagger = MeCab.Tagger()
    tokens = []
    node = tagger.parseToNode(text)
    while node:
        surface = node.surface
        features = node.feature.split(",")
        pos = features[0]  # 品詞情報
        # 不要な数字やトークンを除去する必要がある
        if surface and pos not in {"記号"} and not surface.isdigit():
            tokens.append(surface)
        node = node.next
    return " ".join(tokens)  # PKEが期待する形式はスペース区切りのトークン列
```

```python
# キーワードと検索結果のタイトルが一致する場合のみwikipediaから情報取得
import wikipedia

wikipedia.set_lang("ja")

def get_wikipedia_ver2(keyword):
    try:
        # 検索結果のタイトル一覧を取得
        search_results = wikipedia.search(keyword)

        # 検索結果の中にキーワードと完全一致するタイトルがあるか確認
        if keyword in search_results:
            # 要約文の最初の一文を取得
            summary = wikipedia.summary(keyword).split('。')[0]
            return summary
        else:
            print('検索結果が完全一致しないためwikipediaでは取得しない')
            return '情報なし'

    except wikipedia.DisambiguationError as e:
        print('ページがなく候補のみのためwikipediaでは取得しない')
        return '情報なし'
    except wikipedia.PageError:
        print('ページが見つからないためwikipediaでは取得しない')
        return '情報なし'
    except wikipedia.WikipediaException as e:
        print('wikipediaのエラーのためwikipediaでは取得しない')
        return '情報なし'
    except Exception as e:
        print('エラーが発生したためwikipediaでは取得しない')
        return '情報なし'
```

### seleniumを使用して取得したい情報(クラス)を取得する関数

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
import time
import random

def get_edge(keyword):

    options = Options()
    options.add_argument("Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/132.0.0.0 Safari/537.36 Edg/132.0.0.0")
    
    driver = webdriver.Edge(options=options)

    try:
        # Bing のトップページを開く
        driver.get('https://bing.com')
    
        # 検索ボックスにキーワードを入力して検索を実行
        element = driver.find_element(By.ID, 'sb_form_q')  # 検索ボックス
        element.send_keys(f'{keyword}とは')
        element.submit()  # フォームを送信
    
        # 検索結果が表示されるまでランダムな時間待機
        time.sleep(random.uniform(3, 7))
    
        # b_focusText を含むクラス名の要素を取得
        try:
            element = driver.find_element(By.CSS_SELECTOR, "[class*='b_focusText']")
            return element.text.strip()
        except:
            print("情報がなかったためedgeでは取得しない")
            return '情報なし'
            
    except Exception as e:
        print(e)
        print("エラーが発生したためedgeでは取得しない")
        return '情報なし'
    finally:
        time.sleep(60)
        # WebDriver を終了
        driver.quit()
```

```python
import pandas as pd

# キーワードと説明を対応付けてCSVに保存
def save_keywords_to_csv(keywords, keyword_info, output_file):
    rows = []  # フラットなデータを格納するリスト

    # 外側のリストのインデックスをニュース番号として扱う
    for news_id, (kw_list, info_list) in enumerate(zip(keywords, keyword_info)):
        for keyword, info in zip(kw_list, info_list):
            rows.append({
                "ニュース番号": news_id,
                "キーワード": keyword,
                "説明": info
            })

    # データをデータフレームに変換
    df = pd.DataFrame(rows)

    # CSVファイルに保存
    df.to_csv(output_file, index=False, encoding="utf-8-sig")
    print(f"CSVファイルに保存しました: {output_file}")


output_file = "keywords.csv"
save_keywords_to_csv(keywords, keyword_info, output_file)
```
