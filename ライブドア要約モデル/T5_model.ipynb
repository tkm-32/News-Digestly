### 学習データ:テストデータ=8:2のHold-outで実施

```python
from sklearn.model_selection import ShuffleSplit
import pandas as pd
import numpy as np

# データの読み込み
df = pd.read_csv("output.csv")  # CSVファイルを読み込む
X = np.array(df['本文'])
y = np.array(df['要約'])

ss = ShuffleSplit(n_splits=1, 
                  train_size=0.8,
                  test_size=0.2, 
                  random_state=3)

train_index, test_index = next(ss.split(X))
X_train, X_test = X[train_index], X[test_index]
y_train, y_test = y[train_index], y[test_index]
```

### トークナイズ(前処理)　文字　→　数値

```python
from transformers import T5Tokenizer

# 日本語モデルをロード
tokenizer = T5Tokenizer.from_pretrained('sonoisa/t5-base-japanese')

# トークナイズ関数
def preprocess_data(texts, summaries, tokenizer):
    # T5形式に変換
    inputs = []
    for text in texts:
        # T5は様々なタスクに使用されるためsummarize(要約)と明記しておいた方が良い
        inputs.append("summarize: " + str(text))

    model_inputs = tokenizer(
        inputs, max_length=512, truncation=True, padding="max_length"
    )
    labels = tokenizer(
        summaries, max_length=128, truncation=True, padding="max_length"
    )
    model_inputs["labels"] = labels["input_ids"]
    return model_inputs

# トレーニングデータをトークナイズ
train_data = preprocess_data(X_train, y_train, tokenizer)
# テストデータをトークナイズ
test_data = preprocess_data(X_test, y_test, tokenizer)
```

### PyTorchのデータセットクラスを作成し学習データとテストデータを効率よく扱える形式に変換

```python
import torch
from torch.utils.data import Dataset

class NewsSummaryDataset(Dataset):
    def __init__(self, encodings):
        self.encodings = encodings

    def __len__(self):
        return len(self.encodings["input_ids"])

    def __getitem__(self, idx):
        item = {}
        for key, val in self.encodings.items():
            item[key] = torch.tensor(val[idx])

        return item

# PyTorchデータセットの作成
train_dataset = NewsSummaryDataset(train_data)
test_dataset = NewsSummaryDataset(test_data)
```

**memo**
- 今回使用するtransformersは最新のKerasと対応しておらず、tensowflowと相性が悪い
- 一方でtransformersはPytorchと相性が良い

### モデルの読み込み

```python
from transformers import T5ForConditionalGeneration

# モデルのロード
model = T5ForConditionalGeneration.from_pretrained('sonoisa/t5-base-japanese')
```

### モデルを学習する際の設定を定義

```python
from transformers import TrainingArguments

# トレーニング引数の設定
training_args = TrainingArguments(
    output_dir="./results",          # モデルの保存先
    eval_strategy="epoch",           # 各エポックで評価
    learning_rate=5e-5,              # 学習率
    per_device_train_batch_size=4,   # バッチサイズ
    per_device_eval_batch_size=4,    # 評価時のバッチサイズ
    num_train_epochs=3,              # エポック数
    weight_decay=0.01,               # 正則化
    logging_dir="./logs",            # ログ保存先
)
```

### モデル学習の準備

```python
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
)
```

### 要約モデルを用いておためしで要約文を生成

```python
from transformers import T5Tokenizer, T5ForConditionalGeneration
import pandas as pd

# 保存済みモデルのディレクトリ
model_dir = "./results_ver2"
tokenizer = T5Tokenizer.from_pretrained(model_dir)
model = T5ForConditionalGeneration.from_pretrained(model_dir)

# ニュース本文
input_texts = "大みそか恒例、年末ジャンボ宝くじの抽選会が行われ、1等・前後賞合わせて10億円の当選番号が発表されました。今年の年末ジャンボ宝くじは1等の7億円が23本、前後賞の1億5000万円が46本で、1等と前後賞を合わせると10億円が当たります。100万円以上の当せん本数は「年末ジャンボミニ」も含めると合計1万4403本に上ります。▼1等の当選番号は110組の10万6348番、▼2等・1000万円は52組の15万1307番、49組の13万5439番、139組の16万7124番、65組の13万8164番、60組の12万2015番、46組の11万6185番、60組の11万0493番、84組の12万0749番です。▼3等・100万円は各組共通で10万3265番と16万4169番です。当せん金の受け取りは来年1月7日からとなっています。"

# 要約結果を保存するリスト
summaries = []

# 各ニュース本文を処理
for text in input_texts:
    # トークナイズ(モデルが認識できるように文字データを数値データに変換させる)
    inputs = tokenizer(
        "summarize: " + text,
        return_tensors="pt",
        max_length=1024,
        truncation=True
    )

    # 要約生成
    summary_ids = model.generate(
        inputs["input_ids"],
        max_length=128,
        num_beams=4,    # 4つの候補から最適な出力を選択
        early_stopping=True
    )

    # 要約結果を取得
    # 数値データ　→　文字データ
    summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
    summaries.append(summary)

# 要約結果を DataFrame に追加して保存
print(summaries)

```

```python
import abc

class Tokenizer(abc.ABC):
  """Abstract base class for a tokenizer.

  Subclasses of Tokenizer must implement the tokenize() method.
  """

  @abc.abstractmethod
  def tokenize(self, text):
    raise NotImplementedError("Tokenizer must override tokenize() method")
```

```python
from rouge_score import rouge_scorer
import MeCab  # MeCabを利用

# トークナイザークラス
class MeCabTokenizer(Tokenizer):
    def __init__(self, use_stemmer=False):
        self._stemmer = use_stemmer
        self.tagger = MeCab.Tagger()
        self.wakati = MeCab.Tagger("-Owakati")

    def tokenize(self, text):
        if self._stemmer:
            node = self.tagger.parseToNode(text)
            original_forms = []
            while node:
                feature = node.feature.split(",")
                original_forms.append(feature[6])  # 原型（lemma）
                node = node.next
            return original_forms
        else:
            return self.wakati.parse(text).split()

# トークナイザーを初期化
mecab_tokenizer = MeCabTokenizer()

# 日本語データ
target = "たかしが公園で遊んでいた。"
prediction = "たかしが公園にいたようだ。"

scorer = rouge_scorer.RougeScorer(["rouge1", "rouge2", "rougeL"], use_stemmer=False, tokenizer=mecab_tokenizer)
rouge_score = scorer.score(target, prediction)

# Rougeスコアの計算
rouge_score = scorer.score(target, prediction)

# 結果の表示
for metric, score in rouge_score.items():
    print(f"{metric}: Precision={score.precision:.4f}, Recall={score.recall:.4f}, F1={score.fmeasure:.4f}")
```

### モデルの性能評価

```python
# 保存済みモデルのディレクトリ
model_dir = "./results_ver2"
tokenizer = T5Tokenizer.from_pretrained(model_dir)
model = T5ForConditionalGeneration.from_pretrained(model_dir)
```

```python
# T5モデルで要約を生成
def generate_summary(input_texts, model, tokenizer, max_length=128):
    summaries = []
    for text in input_texts:
        # トークナイズ
        inputs = tokenizer.encode("summarize: " + text, return_tensors="pt", truncation=True, max_length=1024)
        
        # 要約生成
        outputs = model.generate(inputs, max_length=max_length, min_length=10, length_penalty=2.0, num_beams=4, early_stopping=True)
        
        # デコード
        summary = tokenizer.decode(outputs[0], skip_special_tokens=True)
        summaries.append(summary)
    return summaries
```

```python
# Rougeスコア
import abc
from rouge_score import rouge_scorer
import MeCab  

def calc_rougescore(reference, prediction, use_stemmer=False):
    class Tokenizer(abc.ABC):
    
      @abc.abstractmethod
      def tokenize(self, text):
        raise NotImplementedError("Tokenizer must override tokenize() method")
    
    class MeCabTokenizer(Tokenizer):
        def __init__(self, use_stemmer=False):
            self._stemmer = use_stemmer
            self.tagger = MeCab.Tagger()
            self.wakati = MeCab.Tagger("-Owakati")
    
        def tokenize(self, text):
            if self._stemmer:
                node = self.tagger.parseToNode(text)
                original_forms = []
                while node:
                    feature = node.feature.split(",")
                    original_forms.append(feature[6])  # 原型（lemma）
                    node = node.next
                return original_forms
            else:
                return self.wakati.parse(text).split()

    mecab_tokenizer = MeCabTokenizer()
    
    scorer = rouge_scorer.RougeScorer(["rouge1", "rouge2", "rougeL"], use_stemmer=False, tokenizer=mecab_tokenizer)
    rouge_score = scorer.score(reference, prediction)

    return rouge_score
```

```python
# BERTScore
from bert_score import score

def calc_bertscore(reference, prediction, lang="ja"):
    P, R, F1 = score(reference, prediction, lang=lang, verbose=True)

    return F1
```

### 自分で追加した特殊トークンを削除

```python
modified_summaries = [summary.replace("<1>", "").replace("<2>", "").replace("<3>", "") for summary in summaries]
```

```python
modified_y_test = [test.replace("<S1>", "").replace("<S2>","").replace("<S3>","") for test in y_test]
```

### 性能評価を行うコード

```python


# Rougeスコアを格納するリスト
rouge_scores = {"rouge1": [], "rouge2": [], "rougeL": []}

# 各テストデータに対してRougeスコアを計算
for reference, prediction in zip(modified_y_test, modified_summaries):
    scores = calc_rougescore(reference, prediction)
    # それぞれF1値を求めるため.fmeasure(調和平均)を用いる
    rouge_scores["rouge1"].append(scores["rouge1"].fmeasure)
    rouge_scores["rouge2"].append(scores["rouge2"].fmeasure)
    rouge_scores["rougeL"].append(scores["rougeL"].fmeasure)

# Rougeスコアの平均値を計算
average_scores = {}
for metric, values in scores.items():
    average_scores[metric] = np.mean(values)

print(f"Rouge値の平均値 {average_scores}")

# BERTScoreを計算
bert_F1_scores = calc_bertscore(modified_y_test, modified_summaries, lang="ja")
# BERTScoreの平均値を計算
avg_bert_F1 = bert_F1_scores.mean().item()
print(f"BERTSCcoreのF1値の平均値: {avg_bert_F1:.4f}")
```
